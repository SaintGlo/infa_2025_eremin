```python
def fit_curve(results, sizes, func, label):

    times = [np.mean(results['random'][n]['times']) for n in sizes]
    popt, pcov = curve_fit(func, sizes, times, p0=[1e-10])
    '''Подбор коэффициента по МНК и расчёт его погрешности'''
    
    plt.figure(figsize=(8, 5))
    plt.scatter(sizes, times, label='Экспериментальные точки', color='blue')
    plt.plot(sizes, func(sizes, *popt), '--', label=f'Аппроксимация: {label}', color='red')
    plt.title(f'Fitting: {label}')
    plt.xlabel('N')
    plt.ylabel('Время (мс)')
    plt.legend()
    plt.grid(True)
    plt.show()

# Аппроксимация для квадратичных
fit_curve(res_selection, sizes_o_n2, lambda x, c: c * x**2, 'Selection Sort $c \cdot N^2$')
# Аппроксимация для быстрой
fit_curve(res_quick, sizes_o_nlogn, lambda x, c: c * x * np.log2(x), 'QuickSort $c \cdot N \log_2 N$')
```

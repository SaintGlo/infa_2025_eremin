```python
def selection_sort(arr: list[int | float]) -> tuple[list[int | float], int, int]:
    """
    Сортировка выбором.
    Сложность: Время O(N^2), Память O(1) (in-place).
    """
    cmp, swp = 0, 0
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            cmp += 1
            if arr[j] < arr[min_idx]:
                min_idx = j
        if min_idx != i:
            arr[i], arr[min_idx] = arr[min_idx], arr[i]
            swp += 1
    return arr, cmp, swp
    
def stupid_sort(arr: list[int | float]) -> tuple[list[int | float], int, int]:
    """
    Глупая сортировка. При нарушении порядка меняет элементы местами 
    и возвращается в начало массива.
    Сложность: Время O(N^3) в худшем случае, Память O(1) (in-place).
    """
    cmp, swp = 0, 0
    n = len(arr)
    i = 0
    while i < n - 1:
        cmp += 1
        if arr[i] > arr[i + 1]:
            arr[i], arr[i + 1] = arr[i + 1], arr[i]
            swp += 1
            i = 0  # Возврат в начало
        else:
            i += 1
    return arr, cmp, swp
def quicksort_hoare(arr: list[int | float]) -> tuple[list[int | float], int, int]:
    """
    Быстрая сортировка с разбиением Хоара.

    """
    counters = [0, 0]  # [сравнения, перестановки]
    
    def _sort(low: int, high: int):
        if low >= high:
            return
        
        pivot = arr[(low + high) // 2]
        i, j = low, high
        
        while True:
            while arr[i] < pivot:
                counters[0] += 1
                i += 1
            counters[0] += 1  # Учет сравнения, прервавшего цикл
            
            while arr[j] > pivot:
                counters[0] += 1
                j += 1
            counters[0] += 1  # Учет сравнения, прервавшего цикл
            
            if i <= j:
                arr[i], arr[j] = arr[j], arr[i]
                counters[1] += 1
                i += 1
                j -= 1
            else:
                break
                
        _sort(low, j)
        _sort(i, high)
        
     _sort(0, len(arr) - 1)
     return arr, counters[0], counters[1]
```
